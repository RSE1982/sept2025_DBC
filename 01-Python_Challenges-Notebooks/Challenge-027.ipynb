# List Methods Challenge

## Steps

1. Define a variable named **crew** and assign it the value of this list: **["Jean-Luc", "Wesley", "Warf", "Deanna", "William", "Data", "Geordie", "Tasha"]**
2. Add a **print** statement to print the value of **crew** to the console.
3. Using the **pop()** method, remove the last item from the **crew** list.
4. Add another **print** statement to print the value of **crew** to the console.
5. Using the **append()** method, add the string **"Beverly"** to the **crew** list.
6. Add another **print** statement to print the value of **crew** to the console.
7. Using the **extend()** method, add the strings **"Miles"** and **"Guinan"** to the **crew** list.
8. Add another **print** statement to print the value of **crew** to the console.
9. Use the **sort()** method to sort the **crew** list by the **length** of each string, and **reverse** the list so that the longest name is first, and the shortest names are last.
10. Add a final **print** statement to print the value of **crew** to the console.

## My Solution

```python
crew = ["Jean-Luc", "Beverly", "Wesley", "Worf",
        "Deanna", "William", "Data", "Geordie", "Tasha"]
print(crew)

crew.pop()
print(crew)

crew.remove("Beverly")
crew.append("Pulaski")
print(crew)

crew.remove("Pulaski")
crew.append("Beverly")
print(crew)

crew.remove("Wesley")
print(crew)

crew.remove("Jean-Luc")
crew.remove("Beverly")
crew.remove("Worf")
crew.append("Edward")
print(crew)

crew.extend(["Jean-Luc", "Beverly", "Worf"])
crew.remove("Edward")
print(crew)

crew.extend(["Miles", "Guinan", "Ro"])
print(crew)

crew.sort(key=len, reverse=True)
print(crew)
```
