# Reading Data from a File Challenge

## Steps

1. Create a function called **get_content** that takes one parameter called **file**, which will be the name of a file.
2. The **get_content** function should use the correct methods shown in the previous lesson to **open** the file **read** the contents of the **file,** close the file, and **return** the contents from the function.
3. Create a variable **lyrics** and assign it the value of the return from calling the function **get_content** with the argument: **"lyrics.txt"**
4. **Print** the value of the **lyrics** variable to the terminal.

## My Solution

```python
def get_content(file):
    f = open(file, 'r')
    contents = f.read()
    f.close()
    return contents

lyrics = get_content("sources/lyrics-2.txt")
print(lyrics)
```
