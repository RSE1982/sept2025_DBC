# If/Else Challenge

## Steps

1. Write an if-else statement that sets the variable named result to the value specified in the steps below
2. If a is equal to b, the result should be assigned: 'a has the same value as b'
3. Otherwise, the result should be assigned: 'a has not got the same value as b'
4. You may update the variables a and b to test whether your if-else statement is working correctly during development, but they must be set back to pass the tests

```python
a = 10
b = 20
result = None
```

## My Solution

```python
if a == b:
    result = 'a has the same value as b'
else:
    result = 'a has not got the same value as b'

print(result)
```
