# random Challenge

## Steps

1. import random
2. Define a function named **ten_rand_nums (that takes no parameters)** and that returns a **list** of **10** **random** integers between **0** and **100**
3. Call the ten_rand_nums function and assign the returned value to a variable named **result**
4. Print the value of the **result** list to the terminal
5. Use the **shuffle** method of the **random** library to shuffle the **result** list
6. Print the value of the **result** list to the terminal again

## My Solution

```python
import random

def ten_rand_nums():
    rand_nums = []
    for _ in range(10):
        rand_nums.append(random.randint(0, 100))
    return rand_nums

result = ten_rand_nums()
print(result)

random.shuffle(result)
print(result)
```
