# Controlling Iteration with break Challenge

## Steps

1. Create an infinite while loop
2. It should contain an if statement that checks if the variable x is equal to 8 and if so break.
3. Otherwise, it should print out the value of x
4. It should increment the variable x by 1 each time it loops,

## My Solution

```python
x = 0

while True:
    if x == 8:
        break
    print(x)
    x += 1
```
