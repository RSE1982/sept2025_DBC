# List Challenge

## Steps

1. Create a variable named all_numbers and assign it the list value of [1, 5, 44, 22, 13, 17, 56, 3, 88, 9, 97]
2. Create an empty list named big_numbers
3. Create a for loop, to loop through the all_numbers list
4. Inside the for loop, create an if statement to check if the value in the all_numbers list is greater than 40.
5. If the number is larger than 40, add it to the big_numbers list using the append method.
6. Outside the loop: add two print statements to print out the values for all_numbers and big_numbers to the terminal (You will not be tested for this step)

## My Solution

```python
all_numbers = [1, 5, 44, 22, 13, 17, 56, 3, 88, 9, 97]
big_numbers = []

for i in range(len(all_numbers)):
    if all_numbers[i] > 40:
        big_numbers.append(all_numbers[i])
        
print(all_numbers)
print(big_numbers)
```
