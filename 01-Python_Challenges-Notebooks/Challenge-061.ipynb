# Catching Generic Errors With “as” Challenge

## Steps

1. Create a function named: **append_to_list** it should take two parameters named: **ls, val**
2. Inside the Function:
3. Add a **try** block
4. Inside the try block it should append the val to the ls
5. Add a **except** block, which is to catch **AttributeError** as **e**
6. Inside the except block it should print out, either **e.object** or **e.args[0]**, you need to figure out which.
7. After the try except blocks it should return the **ls** variable
8. Outside the function call the **append_to_list** function, passing in the arguments: **values, 4**

## My Solution

```python
values = (10, 5)

def append_to_list(ls, val):
    try:
        ls.append(val)
    except AttributeError as e:
        print(e.args[0])
    return ls

append_to_list(values, 4)


# Do Not Place Code Below This Line 
#this will print out the values variable after the append_to_list function is called
print(values)
```
