# Catching Specific Errors Challenge

## Steps

1. Write a **try, except statement**. In the **try** statement, try to **print** a variable named **non_existent_variable** that has not been defined.
2. Use an **except** statement to check for a **NameError** and **print** **"Variable not defined"** to the terminal if a NameError occurs.

# My Solution

```python
try:
    print(non_existent_variable)
except NameError:
    print("Variable not defined")
```
