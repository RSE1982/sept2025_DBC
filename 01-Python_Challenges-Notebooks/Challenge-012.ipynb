# Determining Truth With Logical Operators Challenge

## Steps

- Define a variable, a, which has a value of 10
- Define a variable, b, which has a value of 5
- Define a variable, result_one, which determines if both a is greater than b and a is greater than 10
- Define a variable, result_two, which determines if either a is equal to 5 or b is less than 5
- Define a variable, result_three, which reverses/negates result_two
- print out the following variables, result_one, result_two, result_three so that you can see their values and to make sure you assigned the correct values to them.

## My Solution

```python
a = 10
b = 5

result_one = (a > b) and (a > 10)
result_two = (a == 5) or (b < 5)
result_three = not(result_two)

print(result_one)
print(result_two)
print(result_three)
```
