# The global Keyword Challenge

## Steps

1. Using what you have learned about python variable scope to fix the code so that the global test_passed variable is updated to True when the code runs, and True is printed to the terminal.

## My Solution

```python
test_passed = False
answer = "friend"

def speak_friend_and_enter():
    global test_passed
    if answer == "friend":
        test_passed = True
        
speak_friend_and_enter()
print(test_passed)
```
