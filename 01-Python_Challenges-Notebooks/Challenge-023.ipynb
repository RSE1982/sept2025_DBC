# Controlling Iteration with pass Challenge

## Steps

1. Create a while loop that runs while the variable x is less than 14
2. It should contain an if-else statement
3. It should check if x > 4 and x < 11
4. If x is between these two limits it should just pass
5. else it should print x
6. After the if-else, it should increment the variable x by 1

## My Solution

```python
x = 0

while x < 14:
    if x>4 and x<11:
        pass
    else:
        print(x)
    x += 1
```
