# Equality vs. Identity Challenge

```python
list_a = [10, 20, 30,]
list_b = [10, 20, 30,]
list_c = list_a
```

## Steps

**Using print()**

1. print out whether list_a **is** list_b
2. print out whether list_a **is equal to** list_b
3. print out whether list_a **is** list_c

## My Solution

```python
print(list_a is list_b)
print(list_a == list_b)
print(list_a is list_c)
```
