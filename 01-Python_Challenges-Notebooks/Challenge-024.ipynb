# Nested Iteration Challenge

## Steps

1. Create a variable **x** and assign it a value: **0**
2. Create a while loop that loops while **x** is less or equal to **3**
3. Within the while loop, add a print statement to print out **x**
4. Within the while loop increment x by 1 on each iteration
5. At this stage, your while loop should print out 0 1 2 3
6. **Remove** the print statement
7. Within the while loop, add another variable **y** assign it a value of: **200** (This has to go above where you increment the **x** variable)
8. Create a while loop that loops while **y** is less or equal to **203**
9. Within this while loop, add: **print(x, y)**
10. Underneath the **print** statement and within the body of this while loop increment **y** by **1** for each iteration of the loop
11. If everything is correct, your **nested** while loop print out should match the result in the image above.

## My Solution

```python
x = 0

while x <= 3:
    y = 200
    while y <= 203:
        print(x,y)
        y+=1
    x+=1
```
