# Incrementing & Decrementing Challenge

## Steps.

1. Create a variable **num** and assign it a value of **100** using the **assignment operator**
2. On the next line type: **print(num)**
3. On the next line reassign to the **num** variable using the **floor assignment operator** and the value **50**
4. On the next line type: **print(num)**
5. On the next line create a variable **num_b** and assign it a value of **100** using the **assignment operator**
6. On the next line type: **print(num_b)**
7. On the next line reassign to the **num_b** variable using the **modulo assignment operator** and the value **3**
8. On the next line type: **print(num_b)**

## My Solution

```python
num = 100
print(num)

num //= 50
print(num)

num_b = 100
print(num_b)

num_b %= 3
print(num_b)
```
