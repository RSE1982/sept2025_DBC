# String Formatting Challenge

## Steps

1. Using **string concatenation**, build the string "Igor is 35" using the **name** and **age** variables, and assign it to a variable named **concat_string**.
2. Write a **print** statement to print the value of **concat_string** to the terminal.
3. Using an **f-string**, build the string "Igor is 35" using the **name** and **age** variables, and assign it to a variable named **f_string**.
4. Write a **print** statement to print the value of **f_string** to the terminal.

```python
name = "Igor"
age = 35
```

## My Solution

```python
concat_string = name + ' is ' + str(age)
print(concat_string)

f_string = f"{name} is {age}"
print(f_string)
```
