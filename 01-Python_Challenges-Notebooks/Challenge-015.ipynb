# Checking Containment with Containment Operators Challenge

## Steps

1. Create a print() statement to check if the string "rat" is in the string "crate"
2. Create a print() statement to check if the string "ink" is NOT in the string "sink"
3. Create a print() statement to check if the string "robbie" is in this list ["gary", "howard", "mark", "jason"]

## My Solution

```python
print("rat" in "crate")
print("ink" not in "sink")
print("robbie" in  ["gary", "howard", "mark", "jason"])
```
