# String Methods Challenge

## Steps

1. Use the **lower()** method on the **dwarves** variable and assign the string to a variable named **lowercase_string**
2. Write a **print** statement to print the value of **lowercase_string** to the terminal
3. Use the **replace()** method on the **lowercase_string** variable to replace all the **commas** in the string with an **empty string**, and assign the string to a variable named **commas_removed**
4. Write a **print** statement to print the value of **commas_removed** to the terminal
5. Use the **split()** method on the **commas_removed** variable, to split the string into a list of **single words,** and assign the list to a variable named **split_list**.
6. Write a **print** statement to print the value of **split_list** to the terminal

## My Solution

```python
dwarves = "Grumpy, Dopey, Doc, Happy, Bashful, Sneezy, Sleepy"
print(dwarves)

lowercase_string = dwarves.lower()
print(dwarves.lower())

commas_removed = lowercase_string.replace(",","")
print(commas_removed)

split_list = commas_removed.split()
print(split_list)
```
