You can read JSON files also
  * You noticed there is `anscombe.json` file
  * Let's read it using `pd.read_json()`. The documentation for this function is [here](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.read_json.html)

* When you are familiar with the JSON file, it is necessary to consider the expected JSON string format. This is done with the argument `orient`. We selected `orient='records'`, which stores the values in this format:
  *  list like [{column -> value}, … , {column -> value}]

```python
df.to_json('outputs/json_example.json', orient='records')
```

### HTML files

image.png

You can get tables from URLs and process them like DataFrames


We are interested in reading the table from the [FIFA World Cup page on Wikipedia](https://en.wikipedia.org/wiki/FIFA_World_Cup) using `pd.read_html()`. You should parse the link.
  * The central aspect of the URL is to contain tables, so the function can read the tables and convert them to DataFrames
  * Please visit the link above so you can get familiar with the content

```python
import requests

url = 'http://en.wikipedia.org/wiki/FIFA_World_Cup'

header = {
  "User-Agent": "Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/50.0.2661.75 Safari/537.36",
  "X-Requested-With": "XMLHttpRequest"
}

r = requests.get(url, headers=header)

search = pd.read_html(r.text)
```

Let's investigate the "`search`" content. You will notice it is not a DataFrame

```python
search
```

You will notice the type of the answer is not a DataFrame, but a Python list

```python
type(search)
```

And using `len()` you can discover how many items this list has

```python
len(search)
```

Each item in the list is a DataFrame. The function `pd.read_html()` got all the tables in that link and transformed them to DataFrames, and placed them in a Python list
* It will be your job to find the DataFrame you are interested in by searching across the list elements
* Let's investigate the first list element

```python
search[0]
```

Let's try the last item

```python
search[24]
```

For this exercise, we will be happy with `search[3]`, showing the soccer world cup editions, indicating the year, hosts and top 4 nations
* You will notice the dataframe needs to be processed to extract meaningful insights. This exercise's objective is to read the table from an URL, not to process the data.
  * For example, the first row is, in fact, the column's name. 
  * The columns with only NaN indicate blank spaces into the original HTML table.

```python
df = search[3]
df
```
